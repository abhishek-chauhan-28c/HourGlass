```python
# Reading the Excel file
wb = load_workbook('/Users/abhishekchauhan/Downloads/Kotak News Aug_updated.xlsx')

# creating the active worksheet object
ws = wb['Mentions']


# Iterate through all cells in column O
for cell in ws['J']:
    # 1. Skip empty cells to prevent errors
    if cell.value is None:
        continue
        
    cell_str = str(cell.value)
    
    # 2. Safely check if the cell contains an '=' formula and double quotes
    if cell_str.startswith('=HYPERLINK') and '"' in cell_str:
        # Extract the URL
        extracted_url = cell_str.split('"')[1]
        
        # 3. Write to Column BI on the EXACT SAME ROW (e.g., BI2, BI3, etc.)
        ws[f'BI{cell.row}'] = extracted_url
        
    elif cell.hyperlink:
        # Fallback: Just in case some cells use standard inserted hyperlinks
        ws[f'BI{cell.row}'] = cell.hyperlink.target

#convert this to pandas dataframe and save it to a new excel file
data  = list(ws.values)
columns = data[0]
rows = data[1:]
df = pd.DataFrame(rows, columns=columns)
df.columns
```

```python
#list_to_remove = ['Combine text', 'Sentiment', 'Cluster', 'Stock', 'Relevant']

# Adding New COlumns to the DataFrame
new_columns = ['Combine text', 'Sentiment', 'Cluster', 'Stock', 'Relevant']

df['Combine text'] = df['Title'].astype(str) + ' ' + df['Text'].astype(str)
df['Sentiment'] = ''
df['Cluster'] = ''
df['Stock'] = ''
df['Relevant'] = ''

# performing the regx 


pattern = re.compile(
    r'\b(?:mutual\s*fund|stocks?|shares?|nifty|sensex|trade|mutualfund|gdp|index|nse|bse|equity)\b'
    r'|शेयर|સેન્સેક્સ|मार्केट|निफ्टी|सेंसेक्स|स्टॉक्स',
    re.IGNORECASE
)

def find_stock_terms(x):
    if pd.isna(x):
        return ''
    return ', '.join(pattern.findall(str(x)))

df['Stock'] = df['Combine text'].apply(find_stock_terms)

 
 
```
